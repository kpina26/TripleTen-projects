## Model analysis

```python
def evaluate_with_timing(model_name, model, X_train, y_train, X_test, y_test):
    start_train = time.time()
    model.fit(X_train, y_train)
    train_time = time.time() - start_train

    start_pred = time.time()
    preds = model.predict(X_test)
    pred_time = time.time() - start_pred

    rmse = np.sqrt(mean_squared_error(y_test, preds))

    return {
        "Model": model_name,
        "RMSE": round(rmse, 2),
        "Training Time (sec)": round(train_time, 2),
        "Prediction Time (sec)": round(pred_time, 4)
    }
```

```python
results = []

# Linear Regression
results.append(evaluate_with_timing("Linear Regression", LinearRegression(), features_train, target_train, features_test, target_test))

# Decision Tree 
results.append(evaluate_with_timing("Decision Tree", dt_grid.best_estimator_, features_train, target_train, features_test, target_test))

# Random Forest
results.append(evaluate_with_timing("Random Forest", rf_grid.best_estimator_, features_train, target_train, features_test, target_test))

# LightGBM
results.append(evaluate_with_timing("LightGBM", lgb_grid.best_estimator_, features_train, target_train, features_test, target_test))
```

<b>   Reviewer's comment </b>


- After we train all models, it is recommended that we choose the best **one** and check its performance on the test subset. Here we only need to make predictions and calculate RMSE. For the final testing, where we use the test subset to check the model's generalization ability, we should use the best model (one model or two models if they have almost the same metric values). We don't use all models here because even just checking their performance influences our choices. This leads to test set leakage when we unconsciously start picking models that perform well on the test set, making it part of the training loop. In real-world scenarios, the test set is meant to reflect how the final model performs in the wild. In practice, you only deploy one model, not several models, so testing just that final one mirrors reality. Moreover, evaluating every tuned model on the test set (especially with big models or datasets) is expensive and time-consuming. 




- When choosing the best model, we have to consider prediction time as well. The best model isn't always the one with the lowest error. Sometimes the errors are only slightly different, but the prediction time varies significantly. In such cases, it's worth considering a faster model. Think of a slow search engine that finds 10 useful links versus a fast one that finds 9. This is especially important if the model needs to operate in real time and produce results repeatedly. If a program runs just once, its speed might not even matter. But if it’s used continuously, optimization becomes crucial. So, in practice, apart from the other requirements, there are also runtime constraints for the model.

</div>

# Conclusion

I developed and evaluated multiple machine learning models to predict the market value of used cars for Rusty Bargain, using historical vehicle data. The goal was to optimize model quality, prediction speed, and training efficiency. I began with data preprocessing, handling missing values, converting date formats, and preparing the dataset for modeling. Several models were trained and tuned which were Linear Regression, Decision Tree, Random Forest, and LightGBM. Linear Regression served as a performance baseline but delivered the highest RMSE and limited predictive power. The Decision Tree model improved accuracy but was slow to train. Random Forest achieved the best predictive accuracy, while LightGBM provided nearly the same quality with much faster training and inference times. After comparing all models based on RMSE, training duration, and prediction time, LightGBM emerged as the most balanced and practical choice for deployment. Overall, this project demonstrated the value of model experimentation and tuning to meet both business and technical requirements for accurate and efficient car price prediction.
