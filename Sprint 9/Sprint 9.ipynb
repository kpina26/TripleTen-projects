# OilyGiant Mining Company

 OilyGiant, aims to identify the most profitable and least risky region for developing a new oil well. Geological data from three regions were analyzed, each containing features (f0, f1, f2) and the target variable product, representing the volume of reserves in thousand barrels. Linear regression model will be trained for each region using a 75:25 training-validation split, with performance evaluated using RMSE. The top 200 wells with the highest predicted reserves will be selected, and their actual reserves are used to calculate revenue and profit based on a $100 million on budget and $4,500 revenue per thousand barrels. A bootstrapping technique with 1,000 iterations will be applied to estimate average profit 95% confidence intervals with the risk of loss. The region with the highest expected profit and a loss risk below 2.5% will be selected as the optimal development site.

In this section, the data for all three regions (geo_data_0.csv, geo_data_1.csv, and geo_data_2.csv) was loaded, inspected using .head(), .info(), .describe(), .duplicated().sum(), and .isnull() to confirm that each dataset contains no missing or duplicate values and is ready for modeling.

# Steps 2-5

```python
datasets = [geo_0, geo_1, geo_2]
```

```python
summary = []
#step 5
def bootstrap_profit(results_df, n_samples=1000):
    profits = []
    wells = 200
    budget = 100_000_000
    revenue_per_unit = 4500

    for _ in range(n_samples):
        sample = results_df.sample(n=500, replace=True)
        top_200 = sample.nlargest(200, 'predicted')
        actual_reserves = top_200['actual'].sum()
        total_revenue = actual_reserves * revenue_per_unit
        profit = total_revenue - budget
        profits.append(profit)

    profits_series = pd.Series(profits)
    mean_profit = profits_series.mean()
    lower = profits_series.quantile(0.025)
    upper = profits_series.quantile(0.975)
    risk = (profits_series < 0).mean() * 100

    return mean_profit, lower, upper, risk
#Even though step 5 was put first this is to calculate risk and profit using bootstrap

for i, dataset in enumerate(datasets):
#This loop iterates over each dataset (representing a region), allowing the same preprocessing, model training, and prediction steps to be applied to all three regions (gd0, gd1, and gd2) efficiently and without code duplication.
 
#step 2
    features = dataset[['f0', 'f1', 'f2']]
    target = dataset['product']
    features_train, features_valid, target_train, target_valid = train_test_split(features, target, test_size=0.25, random_state=42)
    model = LinearRegression()
    model.fit(features_train, target_train)
    predictions = model.predict(features_valid)
    results = pd.DataFrame({'predicted': predictions, 'actual': target_valid})
    rmse = np.sqrt(mean_squared_error(target_valid, predictions))
    average_predicted = predictions.mean()
#This loop allows the code block (data splitting, model training, prediction) to be executed separately for each region.    

#step 3
    revenue_per_unit = 4500
    budget = 100_000_000
    wells = 200
    cost_per_well = budget / wells
    required_reserves = cost_per_well / revenue_per_unit
#This loop allows the code to prepare for profit calculation

#step 4    
    top_200 = results.sort_values(by='predicted', ascending=False).head(200)
    total_actual_reserves = top_200['actual'].sum()
    revenue_per_unit = 4500
    total_revenue = total_actual_reserves * revenue_per_unit
    total_profit = total_revenue - budget
#This loop allows me to calculate profit for each region      
    

    mean_profit, lower, upper, risk = bootstrap_profit(results)
    
    print(f"Region: {i}")
    print(f"Average Predicted Reserves: {average_predicted}")
    print(f"RMSE: {rmse}\n")
    print(f"Required Reserves to Avoid Losses: {required_reserves}")
    print(f"Average Predicted Reserves vs Required Reserves: {'Sufficient' if average_predicted >= required_reserves else 'Insufficient'}")
    print(f"Top 200 total actual reserves: {total_actual_reserves:.2f} thousand barrels")
    print(f"Total Revenue: ${total_revenue:,.2f}")
    print(f"Total Profit: ${total_profit:,.2f}")
    print(f"Bootstrapped Mean Profit: ${mean_profit:,.2f}")
    print(f"95% Confidence Interval: [${lower:,.2f}, ${upper:,.2f}]")
    print(f"Risk of Losses: {risk:.2f}%")
    print('----')
    
    summary.append({'region': i,'mean_profit': mean_profit,'risk': risk,'rmse': rmse})
    safe_regions = [r for r in summary if r['risk'] < 2.5]
if safe_regions:
    best = max(safe_regions, key=lambda x: x['mean_profit'])
    print(f"✅ Best Region to Develop: Region {best['region']}")
    print(f"Expected Profit: ${best['mean_profit']:,.2f}")
    print(f"Risk of Losses: {best['risk']:.2f}%")
else:
    print("⚠️ No region meets the risk threshold (<2.5%)")
```

A bootstrapping approach was implemented to estimate profit distributions by resampling 200 wells from each region 1,000 times. A linear regression model was trained for each region to generate reserve predictions and calculate RMSE. Using the predicted values, the top 200 wells were selected, and total revenue and profit were computed based on actual reserves. The bootstrap function was used to calculate the average profit, 95% confidence interval, and the risk of losses. These metrics were then analyzed to identify the most profitable and least risky region. Region 1 had the highest expected profit and met the risk threshold, the final decision was that region 1 could be recommended for development, as the risk levels across region remained above the acceptable limit.
