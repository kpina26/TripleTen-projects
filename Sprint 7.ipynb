# Split the source data into a training set, a validation set, and a test set

````python
df_train, temp = train_test_split(df_mobile, test_size=0.4, random_state=42)
df_valid, df_test = train_test_split(temp, test_size=0.50, random_state=42)

features = ['calls', 'minutes', 'messages', 'mb_used']

x_train = df_train[features]
y_train = df_train['is_ultra']
x_valid = df_valid[features]
y_valid = df_valid['is_ultra']
x_test = df_test[features]
y_test = df_test['is_ultra']

print(f'x_train shape: {x_train.shape}, y_train shape: {y_train.shape}')
print(f'x_valid shape: {x_valid.shape}, y_valid shape: {y_valid.shape}')
print(f'x_test shape: {x_test.shape}, y_test shape: {y_test.shape}')
````

# Investigate the quality of different models by changing hyperparameters

# Decision Tree Classifier

````python
dt_model = DecisionTreeClassifier(max_depth=5, random_state=42)
dt_model.fit(x_train, y_train)
y_pred_valid = dt_model.predict(x_valid)
valid_accuracy = accuracy_score(y_valid, y_pred_valid)
print(f"Validation Accuracy:", valid_accuracy)
````

````python
for depth in range(1, 11):
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tree.fit(x_train, y_train)
    val_preds = tree.predict(x_valid)
    acc = accuracy_score(y_valid, val_preds)
    print(f"max_depth={depth} | Accuracy: {acc:.3f}")
````

In this part of the project, I explored how changing the max_depth hyperparameter affects the performance of a Decision Tree Classifier. I started by training a model with max_depth=5, which gave me a validation accuracy of around 0.773. To find the best depth, I looped through values from 1 to 10 and trained a new model for each one. I then evaluated each model using the validation set and printed out the accuracy scores. From the results, I observed that the accuracy generally improved as the depth increased, with the best performance at max_depth=8, achieving an accuracy of 0.796. This tuning process helped me choose a depth that avoids underfitting or overfitting.

# Random Forest Classifier

````python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(x_train, y_train)
y_pred = rf_model.predict(x_valid)
accuracy = accuracy_score(y_valid, y_pred)

print(f"Validation Accuracy: {accuracy:.3f}")
````

````python
for n in range(10, 101, 10):
    rf_model = RandomForestClassifier(n_estimators=n, random_state=42)
    rf_model.fit(x_train, y_train)
    y_pred = rf_model.predict(x_valid)
    accuracy = accuracy_score(y_valid, y_pred)
    
    print(f"n_estimators={n} | Accuracy: {accuracy:.3f}")
````

 I evaluated the performance of a Random Forest Classifier by tuning the n_estimators hyperparameter, which controls the number of trees in the forest. I first trained a model with n_estimators=100 and got a validation accuracy of 0.802. I ran a loop from 10 to 100 in steps of 10, training a new model each time. I found that the accuracy gradually improved and started to stabilize around n_estimators=70, with the best results at n_estimators=90 and 100, both reaching an accuracy of 0.802. This experiment helped me confirm that using 100 estimators gives me strong and stable performance without significant overfitting or underfitting.

In machine learning, overfitting occurs when a model learns the training data too well, including its noise and random fluctuations, hindering its ability to generalize to new data. Hyperparameters and model selection significantly impact a model's performance, but tuning them using the test set can lead to data leakage and overfitting.
